```python
def first_task():
    with open("text.txt", "r") as f:
        content = f.read()
        
        print(len(content), len(content.split(" ")), len(content.split("\n")))
```

```python
def second_task():
    biblio = []

    while True:
        inp = input("In: ")

        if inp.lower() == "стоп":
            break

        biblio.append(inp)

    with open("my_books.txt", "w") as f:
        for book in biblio:
            f.write(f"{book}\n")
```

```python
def firth_task():
    century = int(input("In: "))

    print(f"События в {century} веке:")

    with open("history.txt", "r") as f:
        for line in f:
            year = int(line[0: line.find(":")])
            
            if year >= (century - 1) * 100 and year < century * 100:
                print(line.rstrip())
```
