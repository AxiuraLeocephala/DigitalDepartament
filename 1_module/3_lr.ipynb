```python
from random import random
```

```python
def first_task():
    age: int = int(input("In: "))

    if age < 18:
        print("Out: Вы несовершеннолетний")
    elif age >= 18 and age <= 65:
        print("Out: Вы взрослый")
    else:
        print("Out: Вы пенсионер")
```

```python
def second_task():
    mark: int = int(input("In: "))

    if mark > 100 or mark < 0:
        print("Out: Ошибка")

    if mark >= 90:
        print("Out: 5")
    elif mark >= 75:
        print("Out: 4")
    elif mark >= 60:
        print("Out: 3")
    elif mark >= 45:
        print("Out: 2")
    elif mark >= 0:
        print("Out: 1")
```

```python
def third_task():
    num_mounth: int = int(input("In: "))

    if num_mounth in [12, 1, 2]:
        print("Out: Зима")
    elif num_mounth in [3, 4, 5]:
        print("Out: Весна")
    elif num_mounth in [6, 7, 8]:
        print("Out: Лето")
    else:
        print("Out: Осень")
```

```python
LOGIN = "log1"
PASSWORD = "pass1"

def fourth_task(login: str, password: str):
    if login != LOGIN:
        print("Out: Неверный логин")
    else:
        if password != PASSWORD:
            print("Out: Неверный пароль")
        else:
            print("Out: Добро пожаловать")
```

```python
def fifth_task():
    N: int = int(input("In: "))
    sum = 0

    for i in range(0, N+1):
        sum += i

    print("Out: ", sum)
```

```python
def sixth_task():
    random_num: int = random.randint(0, 10)

    while True:
        num = int(input("In: "))

        if num == random_num:
            print("Out: Поздравляю! Вы угадали")
```

```python
def seventh_task():
    num: int = int(input("In: "))

    print(*[f"{num} x {i} = {num * i}\n" for i in range(1, 11)])
```

```python
def eighth_task():
    list_nums: list[int] = []

    while True:
        num = int(input("In: "))

        if num == 0:
            break

        list_nums.append(num)

    sum_list_nums, len_list_nums = sum(list_nums), len(list_nums)    

    print("Out: ", sum_list_nums, len_list_nums, sum_list_nums / len_list_nums)
```

```python
first_task()
second_task()
third_task()
fourth_task("as", "we")
fifth_task()
sixth_task()
seventh_task()
eighth_task()
```
